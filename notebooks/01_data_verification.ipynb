```python
import sys; sys.path.append("..")
from src.data import load_raw

df = load_raw()
print(df.shape)
print(df.columns.tolist())
print(df.isna().sum().sum(), "missing values")
print(df.duplicated().sum(), "duplicate rows")
print(df["default"].value_counts(normalize=True).round(3))

for col in ["EDUCATION", "MARRIAGE", "PAY_0"]:
    print(col, sorted(df[col].unique()))
```

```python
from src.data import clean

df_clean = clean(df)
print(df_clean.shape)
print(df_clean.duplicated().sum(), "duplicate rows remaining")
print("EDUCATION:", sorted(df_clean["EDUCATION"].unique()))
print("MARRIAGE:", sorted(df_clean["MARRIAGE"].unique()))
print("PAY_0:", sorted(df_clean["PAY_0"].unique()))
print(df_clean["default"].value_counts(normalize=True).round(3))
```
